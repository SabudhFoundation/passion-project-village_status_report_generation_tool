```python
import pandas as pd
import warnings

warnings.filterwarnings("ignore")

df = pd.read_excel("All_Villagesof_India_2026-04-10_16-24-30.xlsx", skiprows=1)

print(df.head())
```

```python
print("Columns in dataset:\n")
for col in df.columns:
    print(col)
```

```python
# Clean column names
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ", "_")
df.columns = df.columns.str.replace("(", "", regex=False)
df.columns = df.columns.str.replace(")", "", regex=False)

print("Cleaned Columns:\n")
print(df.columns)
```

```python
df.rename(columns={
    'State_Name_In_English': 'state',
    'District_Name_In_English': 'district',
    'Subdistrict_Name_In_English': 'subdistrict',
    'Village_Name_In_English': 'village',
    'Village_Code': 'village_code'
}, inplace=True)

print("Columns renamed ✅")
```

```python
df = df[df['state'] == 'Punjab']

print("Filtered Punjab Data ✅")
print("Remaining rows:", len(df))
```

```python
districts = ['Bathinda', 'Rupnagar', 'Fatehgarh Sahib']

df = df[df['district'].isin(districts)]

print("Filtered Selected Districts ✅")
print("Remaining rows:", len(df))
```

```python
df = df.drop_duplicates()
df = df.dropna(subset=['village', 'village_code'])

df.reset_index(drop=True, inplace=True)

print("Cleaned Data ✅")
```

```python
# Select correct columns
final_df = df[['state', 'district', 'Sub-District_Name_In_English', 'village', 'village_code']]

# Rename for consistency
final_df.rename(columns={
    'Sub-District_Name_In_English': 'subdistrict'
}, inplace=True)

print("Final Data Preview:")
final_df.head()
```

```python
print(df.columns.tolist())
```

```python
final_df.to_csv("cleaned_punjab_villages.csv", index=False)

print("🎉 File saved successfully!")
```

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# Load GP mapping file
gp_df = pd.read_excel("Village_Gram_Panchayat_Mapping_2026-04-10_16-53-28.xlsx", skiprows=1)

print("GP File Loaded ✅")
print(gp_df.columns)
gp_df.head()
```

```python
# Clean column names
gp_df.columns = gp_df.columns.str.strip()
gp_df.columns = gp_df.columns.str.replace(" ", "_")
gp_df.columns = gp_df.columns.str.replace("(", "", regex=False)
gp_df.columns = gp_df.columns.str.replace(")", "", regex=False)

print("Cleaned Columns:")
print(gp_df.columns)
```

```python
gp_df = gp_df[[
    'Village_Code',
    'Village_Name_In_English',
    'Local_Body_Name_In_English',
    'Subdistrict_Name_In_English',
    'District_Name_In_English'
]]

# Rename columns
gp_df.rename(columns={
    'Village_Code': 'village_code',
    'Village_Name_In_English': 'village',
    'Local_Body_Name_In_English': 'gp_name',
    'Subdistrict_Name_In_English': 'block',
    'District_Name_In_English': 'district'
}, inplace=True)

print("GP Data Selected & Renamed ✅")
gp_df.head()
```

```python
# Remove duplicates
gp_df = gp_df.drop_duplicates()

# Remove null values
gp_df = gp_df.dropna(subset=['village_code'])

# Convert to string (VERY IMPORTANT for merge)
gp_df['village_code'] = gp_df['village_code'].astype(str)

print("GP Data Cleaned ✅")
```

```python
# Convert village_code to string
final_df['village_code'] = final_df['village_code'].astype(str)

print("LGD Data Ready ✅")
```

```python
merged_df = pd.merge(
    final_df,
    gp_df,
    on='village_code',
    how='left'
)

print("🎉 Merge Successful!")
merged_df.head()
```

```python
merged_df = merged_df[[
    'state',
    'district_x',   # from LGD
    'subdistrict',
    'block',
    'gp_name',
    'village_x',
    'village_code'
]]

# Rename columns properly
merged_df.rename(columns={
    'district_x': 'district',
    'village_x': 'village'
}, inplace=True)

print("Final Dataset Ready 🚀")
merged_df.head()
```

```python
merged_df.to_csv("final_village_dataset.csv", index=False)

print("🎯 FINAL FILE SAVED SUCCESSFULLY!")
```
